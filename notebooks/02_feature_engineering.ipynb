# Feature Engineering: Derived Metrics

Build and explore the derived metrics from the README: Possession Value Index, Offensive/Defensive Efficiency, Extra-Man Impact, Clearing Margin.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sys.path.insert(0, str(Path.cwd().parent))

from src.models.feature_engineering import (
    build_derived_features,
    get_feature_columns,
    load_and_prepare,
)

df = load_and_prepare()
print(f"Shape after feature engineering: {df.shape}")
df.head()
```

## Derived Feature Distributions

```python
derived_cols = [
    "possession_value_index",
    "offensive_efficiency",
    "defensive_efficiency",
    "extra_man_impact",
    "clearing_margin",
]
available = [c for c in derived_cols if c in df.columns]
df[available].describe()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for ax, col in zip(axes.flat, available + [None] * (6 - len(available))):
    if col:
        df[col].hist(ax=ax, bins=30, edgecolor="black", alpha=0.7)
        ax.set_title(col)
    else:
        ax.axis("off")
plt.suptitle("Derived Feature Distributions")
plt.tight_layout()
plt.show()
```

## Correlation with Winning (Including Derived)

```python
feature_cols = get_feature_columns(include_derived=True, include_clearing_margin=False)
available = [c for c in feature_cols if c in df.columns]

corrs = df[available + ["winning_percentage"]].corr()["winning_percentage"].drop("winning_percentage", errors="ignore")
corrs = corrs.sort_values(key=abs, ascending=False)
print("Correlation with winning_percentage (incl. derived):")
print(corrs.to_string())
```

```python
corr_matrix = df[available + ["winning_percentage"]].corr()
plt.figure(figsize=(14, 12))
sns.heatmap(corr_matrix, annot=False, cmap="RdBu_r", center=0, vmin=-0.6, vmax=0.6)
plt.title("Feature Correlation Matrix (with Derived Features)")
plt.tight_layout()
plt.show()
```
