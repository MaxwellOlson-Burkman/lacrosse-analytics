# Phase 2: Predictive Modeling

Train Random Forest and Gradient Boosting models to predict winning_percentage. Evaluate with MAE and R2. Export artifacts for the RAG pipeline.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.models.train import run_training
from src.models.evaluate import (
    evaluate_model,
    plot_predicted_vs_actual,
    plot_residuals,
    plot_feature_importance,
)
from src.models.export import export_model, export_team_reports
```

## 1. Train Models

```python
model, X_test, y_test, df, feature_names = run_training()

y_pred = model.predict(X_test)
metrics = evaluate_model(model, X_test, y_test)
print(f"Test MAE: {metrics['mae']:.4f}")
print(f"Test R2:  {metrics['r2']:.4f}")
```

## 2. Evaluation Plots

```python
plot_predicted_vs_actual(y_test, y_pred, title="Predicted vs Actual (Test Set)")
```

```python
plot_residuals(y_test, y_pred, title="Residual Plot (Test Set)")
```

```python
plot_feature_importance(model, feature_names, top_n=15)
```

## 3. Export for RAG Pipeline

```python
paths = export_model(model, feature_names, metrics)
print("Exported:")
for k, v in paths.items():
    print(f"  {k}: {v}")
```

```python
reports_dir = export_team_reports(df, model, feature_names)
print(f"Team reports saved to: {reports_dir}")
print(f"  Count: {len(list(reports_dir.glob('*.txt')))} files")
```
